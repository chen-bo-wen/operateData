解决办法：
如何查看Git\usr\bin在哪里：
1、打开命令行工具（比如Windows 的命令提示符或PowerShell，或者macOS 和Linux 的终端）。
2、输入`which git` 或者`where git` 命令，然后按下回车键。

解决方法：

        引用第三方软件，比如 “ git ”。 工具的 ssh 执行文件，脚本。

        操作： “设置” ---》 “系统”   ---》 “高级系统设置” ---》“环境变量”  ---》“ 系统变量 ”  ----》 “path”   -----》 修改 openssh  引用路径。

        例如：

        原地址： %SYSTEMROOT%\System32\OpenSSH\

        改成：E:\Git\usr\bin

       就可以在cmd 中使用ssh 远程。

        

调试VScode:

Vscode使用ssh 可能还是错误；查看终端日志能找到：

 "remote.SSH.path": C:\Windows\System32\OpenSSH\ssh.exe

路径没有改变，操作如下：

设置 ----》 搜索 remote.ssh, SSH.path ---》 修改成：“E:\Git\usr\bin” 

问题解决。
