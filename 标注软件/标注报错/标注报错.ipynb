当出现如下问题时，检查 imagePath 处的值是否和对应的图片名称一致，如果图片叫 1d.jpg，但是 imagePath 处的值叫 1.jpg 的话，就会出现如下报错。

![1.png](1.png)

批量修改 ImagePath 的代码：

```python
# 修改文件后缀名
import os

path = os.getcwd()

for i in os.listdir(path):
    if i.endswith('.bmp'):
        os.rename(i, i[:-4] + '.jpg')
```

```python
# 批量修改 json 文件中的 imagePath 字段
import json
import os

path = os.getcwd()

for i in os.listdir(path):
    if i.endswith('.json'):
        print("处理文件:", i)
        json_file = os.path.join(path, i)
        index = i[-6]  # 获取倒数第5个字符作为索引
        with open(json_file, "r", encoding="utf-8") as f:
            data = json.load(f)         
  
        print("原始 imagePath:", data.get("imagePath"))

        new_path = f"{i[0:-5]}.jpg"       
        data["imagePath"] = new_path

        print("修改后 imagePath:", data["imagePath"])

        with open(json_file, "w", encoding="utf-8") as f:
            json.dump(data, f, ensure_ascii=False, indent=4)
```
