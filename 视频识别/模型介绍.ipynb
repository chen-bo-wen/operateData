## SlowFast Networks
描述：SlowFast 网络由 Facebook AI 提出，它通过同时使用两个不同速度的网络来处理视频数据，一个网络关注低帧率的 "慢" 特征（Slow Pathway），另一个网络关注高帧率的 "快" 特征（Fast Pathway）。这种方法使得模型能够同时捕捉不同时间尺度的特征。

应用：视频分类、动作识别、动作检测等。

[SlowFast Networks](https://github.com/facebookresearch/SlowFast)
