```python
# split_match_data.py
import shutil
import glob
import random
import os

def train_val():
    image_dir = 'newJPEG'   # JPEG 图像目录
    mask_dir = 'output_masks3'     # PNG 掩模目录
    output_dir = 'splitData'   # 输出目录

    # 创建图像输出目录
    os.makedirs(os.path.join(output_dir, 'train').replace('\\', '/'), exist_ok=True)
    os.makedirs(os.path.join(output_dir, 'val').replace('\\', '/'), exist_ok=True)
    os.makedirs(os.path.join(output_dir, 'test').replace('\\', '/'), exist_ok=True)
    # 创建掩膜输出目录
    os.makedirs(os.path.join(output_dir, 'trainannot').replace('\\', '/'), exist_ok=True)
    os.makedirs(os.path.join(output_dir, 'valannot').replace('\\', '/'), exist_ok=True)
    os.makedirs(os.path.join(output_dir, 'testannot').replace('\\', '/'), exist_ok=True)

    # 获取所有图像和掩模文件
    image_files = [f for f in os.listdir(image_dir) if f.endswith('.jpg')]
    mask_files = [f for f in os.listdir(mask_dir) if f.endswith('.png')]

    # 确保每个图像都有对应的掩模
    image_files = [f for f in image_files if f.replace('.jpg', '.png') in mask_files]

    # 打乱文件顺序
    random.shuffle(image_files)

    # 计算划分索引
    total_images = len(image_files)
    train_end = int(0.8 * total_images)
    val_end = int(0.9 * total_images)

    # 划分数据集
    train_files = image_files[:train_end]
    val_files = image_files[train_end:val_end]
    test_files = image_files[val_end:]

    # 定义函数来复制文件
    def copy_files(file_list, src_dir, dest_dir):
        jpgdest_dir = dest_dir
        maskdest_dir = dest_dir + 'annot'
        for file_name in file_list:
            shutil.copy(os.path.join(src_dir, file_name).replace('\\', '/'), os.path.join(jpgdest_dir, file_name).replace('\\', '/')) # 复制图像
            mask_name = file_name.replace('.jpg', '.png')
            shutil.copy(os.path.join(mask_dir, mask_name).replace('\\', '/'), os.path.join(maskdest_dir, mask_name).replace('\\', '/')) # 复制对应的掩模

    # 复制文件到相应的目录
    copy_files(train_files, image_dir, os.path.join(output_dir, 'train'))
    copy_files(val_files, image_dir, os.path.join(output_dir, 'val'))
    copy_files(test_files, image_dir, os.path.join(output_dir, 'test'))


if __name__ == '__main__':
    train_val()
```
