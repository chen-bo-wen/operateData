```python
    # 调用 increment_path：如果 predict 已存在，则返回 predict1、predict2 … 中的下一个未被占用的路径
    #   - exist_ok=False（默认） → 必须递增
    #   - mkdir=True   → 自动创建目录

import cv2
from ultralytics import YOLO
import os
from pathlib import Path
from ultralytics.utils.files import increment_path

def predict():
    model = YOLO(r'/App/lpl/yolo11/runs/train/exp28/weights/best.pt')
    image_dir = r"/App/lpl/yolo11/FY_2_project/datasets/images/test"
    images = os.listdir(image_dir)  
    print("total images number:", len(images))


    # 调用 increment_path：如果 predict 已存在，则返回 predict1、predict2 … 中的下一个未被占用的路径
    #   - exist_ok=False（默认） → 必须递增
    #   - mkdir=True   → 自动创建目录
    base_output = Path(r"/App/lpl/yolo11/runs/detect") / "predict"
    output_dir = increment_path(base_output, exist_ok=False, mkdir=True)   # exist_ok=False → 必须递增
    print(f"结果将保存至:: {output_dir}")

    for i, image in enumerate(images):
        image_path = os.path.join(image_dir, image)
        print(f"Processing {i+1}/{len(images)}: {image}")
        
        results = model.predict(
            source=image_path,  
            imgsz=1088, 
            conf=0.6, 
            iou=0.5,
            device=0, 
            half=False,
            save=True, 
            save_txt=True,
            save_conf=True, 
            project=output_dir,
            name=f"batch_{i//10}", 
            exist_ok=True
        )
    
        for result in results:
            im_array = result.plot()
            
            output_image_path = os.path.join(output_dir, f"result_{image}")
            cv2.imwrite(output_image_path, im_array)
            print(f"结果已保存到: {output_image_path}")

    print(f"所有图片处理完成！结果保存在: {output_dir}")

if __name__ == '__main__':
    predict()
```
