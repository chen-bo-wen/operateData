下面这段代码有一个问题，就是不会整合到非子文件夹下的文件，如：testFolder/11.jpg ，这个文件就没有遍历到

```python
import os, random, shutil
import glob
import pathlib
from pathlib import Path

out_root   = Path('testFolderNewDataset')

# 1. 创建目录结构
for sub in ['images','labels']:
    (out_root/sub).mkdir(parents=True, exist_ok=True)


rootDir = r'D:\02_数据\FY_3_copy\testFolder'.replace('\\','/')
for root,dirs,files in os.walk(rootDir):
    for dir in dirs:
        # 图像文件
        oldImagesPath = os.path.join(root, dir).replace('\\','/')

        # 标签文件
        oldLablesPath = os.path.join(root, dir).replace('\\','/')

        for file in os.listdir(oldImagesPath):
            if any(file.lower().endswith(ext) for ext in ['.jpg', '.jpeg', '.png']): # 对于所有为图片的文件，不管是哪个层级的
                oldFilePath = os.path.join(oldImagesPath, file).replace('\\','/')  # 原图片文件夹路径
                newImagesPath = os.path.join(out_root, 'images', file).replace('\\','/')  # 新图片文件夹路径
                print('oldFilePath', oldFilePath)
                shutil.copy(oldFilePath, newImagesPath)

        for file in os.listdir(oldLablesPath):
            if any(file.lower().endswith(ext) for ext in ['.json']): # 对于所有为标注json的文件，不管是哪个层级的
                oldFilePath = os.path.join(oldImagesPath, file).replace('\\','/')  # 原标注文件夹路径
                newLablesPath = os.path.join(out_root, 'labels', file).replace('\\','/')  # 新标注文件夹路径
                print('oldFilePath', oldFilePath)
                shutil.copy(oldFilePath, newLablesPath)
```
