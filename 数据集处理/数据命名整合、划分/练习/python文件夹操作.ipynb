python批量修改文件夹下的后缀名：如将文件由 .jpg 后缀修改成 .png 后缀

```python
import os

# 指定需要更改文件的目录
dir_path = 'data/practice' # 相对路径

old_suffix = '.jpg'  # 原文件扩展名
new_suffix = '.png'  # 新文件扩展名

# 遍历该目录下的所有文件和子目录
for root, dirs, files in os.walk(dir_path):
    for file in files:
        # 如果文件是以 定义的old_suffix 结尾
        if file.endswith(old_suffix):
            # 获取文件的完整路径
            file_path = os.path.join(root, file)
            
            # 更改文件的扩展名为.py
            new_file_path = os.path.splitext(file_path)[0] + new_suffix
            
            # 使用os模块的rename方法进行重命名
            os.rename(file_path, new_file_path)

print(f'All {old_suffix} files 修改为了 {new_suffix}.')
```

修改文件名：如将所有文件夹下的图片由 cat.1.jpg 修改成 cat_1.jpg 命名的

```python
import os
import shutil

def iter_files(rootDir):
    for root,dirs,files in os.walk(rootDir):
        for file in files:
            if any(file.lower().endswith(ext) for ext in ['.jpg', '.jpeg', '.png']): # 遍历所有为图片的文件，不管是哪个层级的
                Old_File_Name, file_extension = os.path.splitext(file)  # 分离文件名和扩展名

                old_File_Path = os.path.join(root,file).replace('\\','/') # 旧图片路径

                Old_File_Name_split = Old_File_Name.split('.')
                symbol = "_"
                
                New_File_Name = symbol.join(Old_File_Name_split)
                New_File_Path = os.path.join(root, New_File_Name + file_extension).replace('\\','/')

                if os.path.exists(New_File_Path):
                    print('文件已存在，跳过重命名:', New_File_Path)
                    continue

                os.rename(old_File_Path, New_File_Path)


rootDir = 'data/practice2'

iter_files(rootDir)
```

修改子文件夹名称：如将cat子文件夹修改为333_cat，dog子文件夹修改为 333_dog

```python
import os
import shutil

def iter_files(rootDir):
    for root,dirs,files in os.walk(rootDir):
        for dirname in dirs:
            old_dir = os.path.join(root, dirname)

            new_dir = '333' + '_' + dirname  # 新目录名
            new_dir = os.path.join(root, new_dir)
            os.rename(old_dir, new_dir)


rootDir = 'data/practice3'

iter_files(rootDir)
```

遍历所有文件，将是cat的图片和dog的图片分别存储在两个文件夹里，命名为 cat_combine 和 dog_combine

```python
import os

# 指定需要更改文件的目录
dir_path = 'data/practice5' # 相对路径

old_suffix = '.png'  # 原文件扩展名

# 遍历该目录下的所有文件和子目录
for root, dirs, files in os.walk(dir_path):
    for file in files:
        if file.endswith(old_suffix):
            # 获取文件的完整路径
            file_path = os.path.join(root, file)
            newFolder = 'cat_combine' if 'cat' in file else 'dog_combine'
            new_folder_path = os.path.join(dir_path, newFolder).replace('\\','/')

            # 如果新文件夹不存在，则创建它
            if not os.path.exists(new_folder_path):
                os.makedirs(new_folder_path)

            # 将文件移动到新文件夹
            new_file_path = os.path.join(new_folder_path, file) 
            shutil.move(file_path, new_file_path)
```
