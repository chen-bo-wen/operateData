```python
conda create --name openmmlab python=3.8 -y
conda activate openmmlab
```

[安装CUDA](https://developer.nvidia.com/cuda-toolkit-archive)

![3.png](3.png)

```python
wget https://developer.download.nvidia.com/compute/cuda/11.8.0/local_installers/cuda_11.8.0_520.61.05_linux.run
sudo sh cuda_11.8.0_520.61.05_linux.run

nvcc -V

查看所有CUDA安装： 
ls -l /usr/local | grep cuda

sudo rm /usr/local/cuda # 删除现有符号链接，如果存在的话
sudo ln -sf /usr/local/cuda-11.8 /usr/local/cuda

export PATH=/usr/local/cuda/bin${PATH:+:${PATH}}
export LD_LIBRARY_PATH=/usr/local/cuda/lib64${LD_LIBRARY_PATH:+:${LD_LIBRARY_PATH}}
export CUDA_HOME=/usr/local/cuda
```

[安装cuDNN](https://developer.nvidia.com/rdp/cudnn-archive)

```python
wget -c https://developer.download.nvidia.com/compute/redist/cudnn/v8.7.0/local_installers/11.8/cudnn-linux-x86_64-8.7.0.84_cuda11-archive.tar.xz
# 上面这个链接可以去问AI，即与CUDA版本对应的cuDNN版本的安装包地址是怎样的

xz -dk cudnn-linux-x86_64-8.7.0.84_cuda11-archive.tar.xz
tar -xvf cudnn-linux-x86_64-8.7.0.84_cuda11-archive.tar
cd cudnn-linux-x86_64-8.7.0.84_cuda11-archive
sudo cp ./lib/* /usr/local/cuda-11.8/lib64/
sudo cp ./include/* /usr/local/cuda-11.8/include/
```

[安装 torch torchvision torchaudio](https://pytorch.org/get-started/locally/)

![1.png](1.png)

```python
pip3 install torch torchvision torchaudio --index-url https://download.pytorch.org/whl/cu118
```
