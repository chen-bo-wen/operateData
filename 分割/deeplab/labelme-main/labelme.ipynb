## 转换数据集

将自己的数据放在 data_annotated 文件夹里，这样就能转换为 voc 数据集格式

![3.png](3.png)

通过 labelme2voc.py 文件进行转换，python labelme2voc.py data_annotated data_dataset_selfvoc2 --labels labels.txt，其中 labels.txt 是标签文件，里面存储的是各个标签。
