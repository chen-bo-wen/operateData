出现如：<br>
/bin/python3 /App/cbw/evo2/test/test_evo2.py<br>
Traceback (most recent call last):<br>
  File "/App/cbw/evo2/test/test_evo2.py", line 6, in <module><br>
    import torch<br>
ModuleNotFoundError: No module named 'torch'<br>
需要将/bin/python3这个路径修改为别的路径

修改 launch.json 文件：<br>
将对应的 python 路径改为自己的路径

```python
{
    // 使用 IntelliSense 了解相关属性。 
    // 悬停以查看现有属性的描述。
    // 欲了解更多信息，请访问: https://go.microsoft.com/fwlink/?linkid=830387
    "version": "0.2.0",
    "configurations": [
        {
            "name": "Python: Current File",
            "type": "python",
            "request": "launch",
            "program": "${file}",
            "console": "integratedTerminal",
            "python": "/home/huazhi/anaconda3/envs/evo2c/bin/python"
        },
    ]
}
```

修改 setting.json:

```python
{
    "python.pythonPath": "/home/huazhi/anaconda3/envs/evo2c/bin/python" 
}
```

修改 task.json:

```python
{
    "version": "2.0.0",
    "tasks": [
        {
            "label": "python",
            "type": "shell",
            "command": "/home/huazhi/anaconda3/envs/evo2c/bin/python",
            "args": [
                "${file}"
            ],
            "group": {
                "kind": "build",
                "isDefault": true
            },
            "problemMatcher": [
                "$eslint-compact"
            ]
        }
    ]
}
```
