<a href='#top'>▲ Top</a>

在交叉熵损失的数学定义中，真实标签确实可以理解为隐式的独热编码（one-hot encoding），但在实际代码实现时（如PyTorch），无需显式将标签转换为独热编码，直接用类别索引即可。

![entropy.png](entropy.png)

```python
import torch
from math import log
logits = torch.tensor([[-0.5651, 0.2735, 0.2883, 0.7495, 0.4461],
                       [-0.4567, 0.9265, 0.8855, 0.5502, 1.2194]])
label = torch.tensor([1, 2]) # 标签 [1, 2] 是类别索引，无需手动转换为独热编码。PyTorch内部会高效处理
```

```python
# loss = torch.nn.CrossEntropyLoss(reduction='none')
loss = torch.nn.CrossEntropyLoss()
output = loss(logits, label)
print(output)
```

```python
m = torch.nn.LogSoftmax(dim=-1)
b = m(logits) # 得到归一化后的值
print(b)
nl = torch.nn.NLLLoss() # 实例化对象时，可以选择使用weight,size_average 等参数
n_loss = nl(b,label) # 这里面是要和 NLLLoss 类中forward() 方法中的参数保持一致，也就是需要传入 input 和 target 
print(n_loss)
```

```python
m = torch.nn.Softmax(dim=-1)
Softmax_b = m(logits) # 得到归一化后的值
print(Softmax_b)
Softmax_log = torch.log(Softmax_b) # 取对数
print(Softmax_log)

nl = torch.nn.NLLLoss() # 实例化对象时，可以选择使用weight,size_average 等参数
n_loss = nl(Softmax_log,label) # 这里面是要和 NLLLoss 类中forward() 方法中的参数保持一致，也就是需要传入 input 和 target 
print(n_loss)
```

样本1：Label是1 → 对应概率 0.1906<br>
样本2：Label是2 → 对应概率 0.2265<br>
-log(0.1906) = 1.6573 <br>
-log(0.2265) = 1.4849<br>
loss = (1.6573 + 1.4849)/2 = 1.5711
